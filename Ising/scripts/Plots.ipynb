```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Read the CSV file
data_latticeOrd = np.genfromtxt('../data/inference/latticeOrd_gammaArray_nSamples15000.csv', delimiter=',', names=True)

# Extract the columns
beta_latticeOrd = data_latticeOrd['beta']
MF_latticeOrd_gamma = data_latticeOrd['MF_gamma']
PL_latticeOrd_gamma = data_latticeOrd['PL_gamma']
PLWithLasso_latticeOrd_gamma = data_latticeOrd['PLWithLasso_gamma']

data_rrgOrd = np.genfromtxt('../data/inference/rrgOrd_gammaArray_nSamples15000.csv', delimiter=',', names=True)

# Extract the columns
beta_rrgOrd = data_rrgOrd['beta']
MF_rrgOrd_gamma = data_rrgOrd['MF_gamma']
PL_rrgOrd_gamma = data_rrgOrd['PL_gamma']
PLWithLasso_rrgOrd_gamma = data_rrgOrd['PLWithLasso_gamma']

# Read the CSV file for disordered version
data_latticeDis = np.genfromtxt('../data/inference/latticeDis_gammaArray_nSamples15000.csv', delimiter=',', names=True)

# Extract the columns
beta_latticeDis = data_latticeDis['beta']
MF_latticeDis_gamma = data_latticeDis['MF_gamma']
PL_latticeDis_gamma = data_latticeDis['PL_gamma']
PLWithLasso_latticeDis_gamma = data_latticeDis['PLWithLasso_gamma']

data_rrgDis = np.genfromtxt('../data/inference/rrgDis_gammaArray_nSamples15000.csv', delimiter=',', names=True)

# Extract the columns
beta_rrgDis = data_rrgDis['beta']
MF_rrgDis_gamma = data_rrgDis['MF_gamma']
PL_rrgDis_gamma = data_rrgDis['PL_gamma']
PLWithLasso_rrgDis_gamma = data_rrgDis['PLWithLasso_gamma']
```

# Gamma arrays

```python
from matplotlib import rcParams
rcParams['font.family'] = 'serif'
rcParams['font.serif'] = ['Computer Modern']
rcParams['font.size'] = 18
rcParams['text.usetex'] = True

fig, ax = plt.subplots(2, 2, figsize=(10, 8))

# Ordered lattice
ax[0, 0].scatter(beta_latticeOrd, MF_latticeOrd_gamma, zorder=3, color="teal", label="Mean Field")
ax[0, 0].scatter(beta_latticeOrd, PL_latticeOrd_gamma, zorder=3, color="firebrick", label="Pseudo-Likelihood", marker="^")
ax[0, 0].scatter(beta_latticeOrd, PLWithLasso_latticeOrd_gamma, zorder=3, color="forestgreen", label="PL with lasso", marker="s")
ax[0, 0].vlines(1/2.269, 0.01, 10, color="black", ls="--", label=r"$\beta_c$")
ax[0, 0].legend(loc="upper left", fancybox=True, framealpha=1, fontsize=14)
ax[0, 0].set_ylabel(r"$\gamma_J$")
ax[0, 0].set_xlabel(r"$\beta$")
ax[0, 0].set_title("Ordered lattice")
ax[0, 0].grid()
ax[0, 0].set_yscale("log")
ax[0, 0].set_ylim([0.05, 15])
ax[0, 0].set_xlim([0.0, 0.61])

# Ordered Erdős–Rényi
ax[0, 1].scatter(beta_rrgOrd, MF_rrgOrd_gamma, zorder=3, color="teal", label="Mean Field")
ax[0, 1].scatter(beta_rrgOrd, PL_rrgOrd_gamma, zorder=3, color="firebrick", label="Pseudo-Likelihood", marker="^")
ax[0, 1].scatter(beta_rrgOrd, PLWithLasso_rrgOrd_gamma, zorder=3, color="forestgreen", label="PL with lasso", marker="s")
ax[0, 1].vlines(1/3.9152, 0.01, 10, color="black", ls="--", label=r"$\beta_c$")
ax[0, 1].legend(loc="upper left", fancybox=True, framealpha=1, fontsize=14)
ax[0, 1].set_ylabel(r"$\gamma_J$")
ax[0, 1].set_xlabel(r"$\beta$")
ax[0, 1].set_title("Ordered Erdős–Rényi")
ax[0, 1].grid()
ax[0, 1].set_yscale("log")
ax[0, 1].set_ylim([0.05, 15])
ax[0, 1].set_xlim([0, 0.61])

# Disordered lattice
ax[1, 0].scatter(beta_latticeDis, MF_latticeDis_gamma, zorder=3, color="teal", label="Mean Field")
ax[1, 0].scatter(beta_latticeDis, PL_latticeDis_gamma, zorder=3, color="firebrick", label="Pseudo-Likelihood", marker="^")
ax[1, 0].scatter(beta_latticeDis, PLWithLasso_latticeDis_gamma, zorder=3, color="forestgreen", label="PL with lasso", marker="s")
#ax[1, 0].vlines(1/2.269, 0.01, 10, color="black", ls="--", label=r"$\beta_c$")
ax[1, 0].legend(loc="upper left", fancybox=True, framealpha=1, fontsize=14)
ax[1, 0].set_ylabel(r"$\gamma_J$")
ax[1, 0].set_xlabel(r"$\beta$")
ax[1, 0].set_title("Disordered lattice")
ax[1, 0].grid()
ax[1, 0].set_yscale("log")
ax[1, 0].set_ylim([0.05, 15])
ax[1, 0].set_xlim([0.0, 0.61])

# Disordered Erdős–Rényi
ax[1, 1].scatter(beta_rrgDis, MF_rrgDis_gamma, zorder=3, color="teal", label="Mean Field")
ax[1, 1].scatter(beta_rrgDis, PL_rrgDis_gamma, zorder=3, color="firebrick", label="Pseudo-Likelihood", marker="^")
ax[1, 1].scatter(beta_rrgDis, PLWithLasso_rrgDis_gamma, zorder=3, color="forestgreen", label="PL with lasso", marker="s")
ax[1, 1].vlines(1/2.85878, 0.01, 10, color="black", ls="--", label=r"$\beta_c$")
ax[1, 1].legend(loc="upper left", fancybox=True, framealpha=1, fontsize=14)
ax[1, 1].set_ylabel(r"$\gamma_J$")
ax[1, 1].set_xlabel(r"$\beta$")
ax[1, 1].set_title("Disordered Erdős–Rényi")
ax[1, 1].grid()
ax[1, 1].set_yscale("log")
ax[1, 1].set_ylim([0.05, 15])
ax[1, 1].set_xlim([0, 0.61])

# Increase font sizes and tick sizes
for row in ax:
    for axis in row:
        axis.title.set_fontsize(20)
        axis.xaxis.label.set_fontsize(16)
        axis.yaxis.label.set_fontsize(16)
        axis.tick_params(axis='both', which='major', labelsize=16)
        axis.tick_params(axis='both', which='minor', labelsize=16)

plt.tight_layout()
plt.savefig("../imgs/gamma_plots.png")
plt.show()
```
